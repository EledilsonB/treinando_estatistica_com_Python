```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dados = {
    'Produto': ['Abacate', 'Banana', 'Maçã', 'Laranja', 'Abacaxi', 'Uva', 'Melão'],
    'Quantidade': [3, 4, 5, 6, 7, 7, 10]
}

df = pd.DataFrame(dados)
df
```

```python
media = df['Quantidade'].mean()
print(media)

media_2 = sum(df['Quantidade']) / len(df['Quantidade'])
print(media_2)
```

```python
mediana = df['Quantidade'].median()
print(mediana)
```

```python
moda = df['Quantidade'].mode()
print(moda)
```

```python
# Verificando a Variancia dos dados

# PARAMETRO DDOF = 0 VARIANCIA POPULACIONAL
# PARAMETRO DDOF = 1 VARIANCIA AMOSTRAL

variancia = df['Quantidade'].var(ddof=0)
print(variancia)
```

```python
# Verificando o DESVIO PADRÃO dos dados

desvio_padrao = df['Quantidade'].std(ddof=0) # Desvio padrão populacional.
print(desvio_padrao)
```

```python
# Verificação dos QUARTIS do DataFrame.

print(f'Primeiro Quartil: {df['Quantidade'].quantile(q=.25)}')
print(f'Segundo Quartil: {df['Quantidade'].quantile(q=.5)}')
print(f'Terceiro Quartil: {df['Quantidade'].quantile(q=.75)}')
print(f'Amplitude: {df['Quantidade'].max() - df['Quantidade'].min()}')

```

```python
dados = [3, 4, 5, 6, 7, 7, 10]
media = np.mean(dados)
x = range(0, len(dados))

plt.scatter(x, dados)
plt.axhline(y = media, color='r', linestyle='-')
plt.ylabel('Preços')
plt.xlabel('Frutas')
plt.title('Histplot distância da média')
plt.show()
```
